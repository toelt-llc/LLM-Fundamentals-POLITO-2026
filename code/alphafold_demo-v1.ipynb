# How are LLM ideas used in AlphaFold?

AlphaFold is not literally ChatGPT for proteins.

However, AlphaFold uses several ideas that are also fundamental in modern Large Language Models (LLMs), especially:

- transformers
- attention mechanisms
- embeddings
- sequence modeling

The key conceptual bridge is:

```text
LLMs learn relationships between words.
AlphaFold learns relationships between amino acids.
```

## Language vs Proteins

In natural language processing (NLP), a sentence is a sequence of words.

Example:

```text
"The cat sits on the mat"
```

In biology, a protein is a sequence of amino acids.

Example:

```text
MKTFFVAGVILALAG
```

Both are sequences.

This means many ideas from language models can also be applied to proteins.

## What transformers learn in language

A transformer-based LLM learns relationships between words.

Example:

```text
"The animal didn't cross the street because it was tired."
```

The model learns that:

```text
"it" → refers to → "animal"
```

This is possible because transformers use an attention mechanism.

## What AlphaFold learns

AlphaFold also processes sequences.

But instead of words, it processes amino acids.

The model learns:
- which amino acids interact
- which residues become close in 3D space
- how proteins fold

Example:

```text
Residue 15 may interact with residue 220
```

even though they are far apart in the sequence.

## Attention mechanisms

Attention is one of the key ideas behind transformers.

The attention mechanism helps the model determine:

```text
Which parts of the sequence are important for understanding another part?
```

This works extremely well for:
- language
- proteins
- DNA sequences
- biological structures

## Embeddings

LLMs convert words into embeddings.

Example:

```text
king → vector
queen → vector
```

Words with similar meaning tend to have similar embeddings.

Protein models do something similar:

```text
protein sequence → embedding vector
```

Proteins with similar biological properties often produce similar embeddings.

## Sequence → Structure

AlphaFold learns the relationship:

```text
protein sequence → 3D structure
```

This is one of the most important problems in biology because protein function depends strongly on structure.

````python
#!pip install py3Dmol
````

````python
import requests
import py3Dmol
````

````python
uniprot_id = "P69905"
````

````python
url = f"https://alphafold.ebi.ac.uk/api/prediction/{uniprot_id}"

response = requests.get(url)

data = response.json()

print(data[0]["entryId"])
print(data[0]["pdbUrl"])
````

````python
pdb_url = data[0]["pdbUrl"]

pdb_text = requests.get(pdb_url).text

print(pdb_text[:500])
````

````python
viewer = py3Dmol.view(width=800, height=500)

viewer.addModel(pdb_text, "pdb")

viewer.setStyle({"cartoon": {"color": "spectrum"}})

viewer.zoomTo()

viewer.show()
````

## What happened?

We used AlphaFold's public database to retrieve a predicted protein structure.

The AI model predicted the 3D arrangement of the protein from its amino-acid sequence.

This is important because protein function depends strongly on structure.

Applications include:
- drug discovery
- enzyme engineering
- antibody design
- disease research
